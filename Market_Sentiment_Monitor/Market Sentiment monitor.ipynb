```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

def analyze_shanghai_sentiment(years=10, stock_symbol=Sentiment_stock):
    
    print(f"=== {stock_symbol} SENTIMENT ANALYSIS ===")
    print(f"Analysis Period: {years} year(s)")
    print(f"Stock Symbol: {stock_symbol}")
    print("=" * 60)
    
    # Configuration
    end_date = datetime.now()
    start_date = end_date - timedelta(days=int(years * 365))
    
    print(f"\nFetching data from {start_date.strftime('%Y-%m-%d')} to {end_date.strftime('%Y-%m-%d')}")
    
    try:
        # Download stock data
        stock_data = yf.download(stock_symbol, start=start_date, end=end_date, progress=False)
        
        if len(stock_data.columns.levels[0]) > 1:
            stock_data = stock_data.droplevel('Ticker', axis=1)  # Clean column structure
        
        stock_data.columns = ['Close', 'High', 'Low', 'Open', 'Volume']
        
        print(f"Successfully downloaded {len(stock_data)} data points")
        
        if len(stock_data) < 50:
            print("Warning: Insufficient data points for reliable analysis")
            return None
            
    except Exception as e:
        print(f"Error downloading data: {e}")
        return None
    
    # Calculate basic price metrics
    stock_data['Daily_Return'] = stock_data['Close'].pct_change()
    stock_data['Price_Change'] = stock_data['Close'].diff()
    
    # 1. Rolling Volatility (20-day annualized)
    stock_data['Volatility_20d'] = stock_data['Daily_Return'].rolling(window=20).std() * np.sqrt(252)
    
    # 2. RSI (Relative Strength Index)
    def calculate_rsi(prices, window=14):
        delta = prices.diff()
        gain = (delta.where(delta > 0, 0)).rolling(window=window).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(window=window).mean()
        rs = gain / loss
        rsi = 100 - (100 / (1 + rs))
        return rsi
    
    stock_data['RSI'] = calculate_rsi(stock_data['Close'])
    
    # 3. MACD (Moving Average Convergence Divergence)
    exp1 = stock_data['Close'].ewm(span=12).mean()
    exp2 = stock_data['Close'].ewm(span=26).mean()
    stock_data['MACD'] = exp1 - exp2
    stock_data['MACD_Signal'] = stock_data['MACD'].ewm(span=9).mean()
    
    # 4. Price relative to moving averages
    stock_data['MA_20'] = stock_data['Close'].rolling(window=20).mean()
    stock_data['MA_50'] = stock_data['Close'].rolling(window=50).mean()
    stock_data['Price_vs_MA20'] = (stock_data['Close'] / stock_data['MA_20'] - 1) * 100
    stock_data['Price_vs_MA50'] = (stock_data['Close'] / stock_data['MA_50'] - 1) * 100
    
    # 5. Volume-based sentiment
    stock_data['Volume_MA'] = stock_data['Volume'].rolling(window=20).mean()
    stock_data['Relative_Volume'] = stock_data['Volume'] / stock_data['Volume_MA']
    
    # 6. Create composite sentiment score
    def normalize_series(series):
        """Normalize series to 0-1 range"""
        clean_series = series.dropna()
        if len(clean_series) == 0:
            return series
        return (series - clean_series.min()) / (clean_series.max() - clean_series.min())
    
    # Normalize indicators
    sentiment_components = pd.DataFrame(index=stock_data.index)
    sentiment_components['RSI_norm'] = normalize_series(stock_data['RSI'])
    sentiment_components['MACD_norm'] = normalize_series(stock_data['MACD'])
    sentiment_components['Price_MA20_norm'] = normalize_series(stock_data['Price_vs_MA20'])
    sentiment_components['Price_MA50_norm'] = normalize_series(stock_data['Price_vs_MA50'])
    
    # Inverse volatility (lower volatility = more positive sentiment)
    sentiment_components['Volatility_inv_norm'] = 1 - normalize_series(stock_data['Volatility_20d'])
    
    # Volume sentiment (bell curve around 1.5x average volume)
    volume_sentiment = np.exp(-((stock_data['Relative_Volume'] - 1.5)**2) / 0.5)
    sentiment_components['Volume_sentiment_norm'] = normalize_series(volume_sentiment)
    
    # Composite sentiment score (equally weighted, scaled 0-100)
    stock_data['Sentiment_Score'] = sentiment_components.mean(axis=1) * 100
    
    print("Sentiment calculation completed!")
    
    # Create the dual-axis visualization with color-coded sentiment
    fig, ax1 = plt.subplots(figsize=(16, 10))
    
    # Plot price on left axis
    color_price = 'tab:blue'
    ax1.set_xlabel('Date', fontsize=12)
    ax1.set_ylabel(' Price', color=color_price, fontsize=12)
    line1 = ax1.plot(stock_data.index, stock_data['Close'], color=color_price, linewidth=2, label='Close Price')
    ax1.tick_params(axis='y', labelcolor=color_price)
    ax1.grid(True, alpha=0.3)
    
    # Create second y-axis for sentiment with gradient colors
    ax2 = ax1.twinx()
    ax2.set_ylabel('Market Sentiment Score (0-100)', fontsize=12)
    
    # Create color-coded sentiment line (light red for panic, light green for greed)
    sentiment_values = stock_data['Sentiment_Score'].values
    dates = stock_data.index
    
    # Create segments with different colors based on sentiment levels
    for i in range(len(sentiment_values)-1):
        if not pd.isna(sentiment_values[i]) and not pd.isna(sentiment_values[i+1]):
            current_sentiment = sentiment_values[i]
            
            # Color mapping based on sentiment levels
            if current_sentiment >= 70:  # Extreme greed
                color = '#FFB6C1'  # Red
            elif current_sentiment >= 60:  # Optimism
                color = '#FFA07A'  # Red
            elif current_sentiment >= 40:  # Neutral
                color = '#FFD700'  # Gold
            elif current_sentiment >= 30:  # Fear
                color = '#98FB98'  # Green
            else:  # Extreme fear/panic
                color = '#90EE90'  # Green
            
            ax2.plot([dates[i], dates[i+1]], [sentiment_values[i], sentiment_values[i+1]], 
                    color=color, linewidth=2, alpha=0.8)
    
    # Add a single legend entry for sentiment
    ax2.plot([], [], color='#90EE90', linewidth=2, label='Sentiment Score (Color-coded)')
    ax2.tick_params(axis='y')
    
    # Add title and legend
    plt.title(f'{stock_symbol}: Price vs Market Sentiment ({years} Year Analysis)', 
              fontsize=16, fontweight='bold', pad=20)
    
    # Combine legends
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    # Add color legend for sentiment
    from matplotlib.patches import Rectangle
    legend_elements = [
        Rectangle((0,0),1,1, facecolor='#90EE90', label='Extreme Fear (<30)'),
        Rectangle((0,0),1,1, facecolor='#98FB98', label='Fear (30-40)'),
        Rectangle((0,0),1,1, facecolor='#FFD700', label='Neutral (40-60)'),
        Rectangle((0,0),1,1, facecolor='#FFA07A', label='Optimism (60-70)'),
        Rectangle((0,0),1,1, facecolor='#FFB6C1', label='Extreme Greed (>70)')
    ]
    ax2.legend(handles=legend_elements, loc='upper center', title='Sentiment Levels')
    
    # Format x-axis
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    # Analysis and Statistics
    print(f"\n=== ANALYSIS RESULTS ({years} YEAR PERIOD) ===")
    
    # Remove NaN values for correlation calculation
    clean_data = stock_data[['Close', 'Sentiment_Score']].dropna()
    correlation = clean_data['Close'].corr(clean_data['Sentiment_Score'])
    
    current_price = stock_data['Close'].iloc[-1]
    current_sentiment = stock_data['Sentiment_Score'].iloc[-1]
    
    print(f"Price-Sentiment Correlation: {correlation:.3f}")
    print(f"Current Price: {current_price:.2f}")
    print(f"Current Sentiment: {current_sentiment:.1f}/100")
    
    # Calculate period return
    period_return = ((stock_data['Close'].iloc[-1] - stock_data['Close'].iloc[0]) / stock_data['Close'].iloc[0]) * 100
    print(f"{years}-Year Return: {period_return:+.1f}%")
    
    # Annualized return
    annualized_return = ((stock_data['Close'].iloc[-1] / stock_data['Close'].iloc[0]) ** (1/years) - 1) * 100
    print(f"Annualized Return: {annualized_return:+.1f}%")
    
    # Price statistics
    price_high = stock_data['Close'].max()
    price_low = stock_data['Close'].min()
    
    print(f"Price vs {years}-Year High: {((current_price - price_high) / price_high) * 100:+.1f}%")
    print(f"Price vs {years}-Year Low: {((current_price - price_low) / price_low) * 100:+.1f}%")
    
    # Sentiment statistics
    sentiment_high = stock_data['Sentiment_Score'].max()
    sentiment_low = stock_data['Sentiment_Score'].min()
    sentiment_avg = stock_data['Sentiment_Score'].mean()
    
    print(f"Sentiment - Current: {current_sentiment:.1f} | Average: {sentiment_avg:.1f} | Range: {sentiment_low:.1f}-{sentiment_high:.1f}")
    
    # Sentiment interpretation
    if current_sentiment >= 70:
        sentiment_label = "Very Bullish (Extreme Greed)"
    elif current_sentiment >= 60:
        sentiment_label = "Bullish (Optimism)"
    elif current_sentiment >= 40:
        sentiment_label = "Neutral (Mixed Sentiment)"
    elif current_sentiment >= 30:
        sentiment_label = "Bearish (Fear)"
    else:
        sentiment_label = "Very Bearish (Extreme Fear)"
    
    print(f"Current Sentiment Level: {sentiment_label}")
    
    # Key events analysis
    print(f"\n=== KEY EVENTS (PAST {years} YEAR(S)) ===")
    stock_data['price_change_pct'] = stock_data['Close'].pct_change() * 100
    
    # Major moves
    major_rallies = stock_data[stock_data['price_change_pct'] > 5]
    major_declines = stock_data[stock_data['price_change_pct'] < -5]
    
    print(f"Total Major Rally Days (>5%): {len(major_rallies)}")
    print(f"Total Major Decline Days (>5%): {len(major_declines)}")
    
    if len(major_rallies) > 0:
        print("\nRecent Major Rally Days (last 5):")
        for date, row in major_rallies.tail(5).iterrows():
            print(f"  {date.strftime('%Y-%m-%d')}: +{row['price_change_pct']:.1f}% | Sentiment: {row['Sentiment_Score']:.1f}")
    
    if len(major_declines) > 0:
        print("\nRecent Major Decline Days (last 5):")
        for date, row in major_declines.tail(5).iterrows():
            print(f"  {date.strftime('%Y-%m-%d')}: {row['price_change_pct']:.1f}% | Sentiment: {row['Sentiment_Score']:.1f}")
    
    # Sentiment extremes
    high_sentiment_days = stock_data[stock_data['Sentiment_Score'] > 70]
    low_sentiment_days = stock_data[stock_data['Sentiment_Score'] < 30]
    
    print(f"\nHigh Sentiment Days (>70): {len(high_sentiment_days)}")
    print(f"Low Sentiment Days (<30): {len(low_sentiment_days)}")
    
    if len(high_sentiment_days) > 0:
        print(f"\nRecent High Sentiment Days (last 5):")
        for date, row in high_sentiment_days.tail(5).iterrows():
            print(f"  {date.strftime('%Y-%m-%d')}: Sentiment {row['Sentiment_Score']:.1f} | Price: {row['Close']:.2f}")
    
    if len(low_sentiment_days) > 0:
        print(f"\nRecent Low Sentiment Days (last 5):")
        for date, row in low_sentiment_days.tail(5).iterrows():
            print(f"  {date.strftime('%Y-%m-%d')}: Sentiment {row['Sentiment_Score']:.1f} | Price: {row['Close']:.2f}")
    
    print(f"\n=== SENTIMENT METHODOLOGY ===")
    print("Composite score combining 6 technical indicators:")
    print("• RSI (Relative Strength Index) - Momentum oscillator")
    print("• MACD (Moving Average Convergence Divergence) - Trend indicator")  
    print("• Price vs 20-day Moving Average - Short-term trend")
    print("• Price vs 50-day Moving Average - Medium-term trend")
    print("• Inverted Volatility - Market stability (lower volatility = higher sentiment)")
    print("• Volume Sentiment - Trading activity patterns")
    print("\nScore Interpretation & Colors:")
    print("• 0-30: Very Bearish (Extreme Fear) - Light Pink")
    print("• 30-40: Bearish (Fear) - Light Salmon")
    print("• 40-60: Neutral (Mixed Sentiment) - Gold")
    print("• 60-70: Bullish (Optimism) - Pale Green")
    print("• 70-100: Very Bullish (Extreme Greed) - Light Green")
    
    # Return results for further analysis
    results = {
        'data': stock_data,
        'correlation': correlation,
        'current_price': current_price,
        'current_sentiment': current_sentiment,
        'period_return': period_return,
        'annualized_return': annualized_return,
        'sentiment_label': sentiment_label,
        'analysis_period': years,
        'major_rallies': len(major_rallies),
        'major_declines': len(major_declines),
        'high_sentiment_days': len(high_sentiment_days),
        'low_sentiment_days': len(low_sentiment_days)
    }
    
    return results

# Usage Examples:
if __name__ == "__main__":
    print(" Sentiment Analysis Tool")
    print("=" * 50)
    
    # Example 1: Default 10-year analysis
    print("\n1. Running default 10-year analysis...")
    results_10y = analyze_shanghai_sentiment()
    
    # Example 2: 5-year analysis
    print("\n" + "="*80)
    print("2. Running 5-year analysis...")
    results_5y = analyze_shanghai_sentiment(years=5)
    
    # Example 3: 2-year analysis
    print("\n" + "="*80)
    print("3. Running 2-year analysis...")
    results_2y = analyze_shanghai_sentiment(years=2)

# Quick function for single analysis
def quick_analysis(years=10):
    """Quick analysis with minimal output"""
    return analyze_shanghai_sentiment(years=years)

# For manual adjustment, use:
# years = 5  # Set desired number of years
# results = analyze_shanghai_sentiment(years=years)
```

```python
 # Multiple price sentiment
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import warnings
import seaborn as sns
from zoneinfo import ZoneInfo
warnings.filterwarnings('ignore')

def calculate_sentiment_score(stock_data):
    """
    Calculate composite sentiment score for a given stock dataset
    Returns the latest sentiment score or NaN if calculation fails
    """
    try:
        if len(stock_data) < 50:  # Insufficient data
            return np.nan
            
        # Calculate basic price metrics
        stock_data = stock_data.copy()
        stock_data['Daily_Return'] = stock_data['Close'].pct_change()
        
        # 1. Rolling Volatility (20-day annualized)
        stock_data['Volatility_20d'] = stock_data['Daily_Return'].rolling(window=20).std() * np.sqrt(252)
        
        # 2. RSI (Relative Strength Index)
        def calculate_rsi(prices, window=14):
            delta = prices.diff()
            gain = (delta.where(delta > 0, 0)).rolling(window=window).mean()
            loss = (-delta.where(delta < 0, 0)).rolling(window=window).mean()
            rs = gain / loss
            rsi = 100 - (100 / (1 + rs))
            return rsi
        
        stock_data['RSI'] = calculate_rsi(stock_data['Close'])
        
        # 3. MACD (Moving Average Convergence Divergence)
        exp1 = stock_data['Close'].ewm(span=12).mean()
        exp2 = stock_data['Close'].ewm(span=26).mean()
        stock_data['MACD'] = exp1 - exp2
        
        # 4. Price relative to moving averages
        stock_data['MA_20'] = stock_data['Close'].rolling(window=20).mean()
        stock_data['MA_50'] = stock_data['Close'].rolling(window=50).mean()
        stock_data['Price_vs_MA20'] = (stock_data['Close'] / stock_data['MA_20'] - 1) * 100
        stock_data['Price_vs_MA50'] = (stock_data['Close'] / stock_data['MA_50'] - 1) * 100
        
        # 5. Volume-based sentiment
        stock_data['Volume_MA'] = stock_data['Volume'].rolling(window=20).mean()
        stock_data['Relative_Volume'] = stock_data['Volume'] / stock_data['Volume_MA']
        
        # 6. Create composite sentiment score
        def normalize_series(series):
            """Normalize series to 0-1 range"""
            clean_series = series.dropna()
            if len(clean_series) == 0:
                return series
            return (series - clean_series.min()) / (clean_series.max() - clean_series.min())
        
        # Normalize indicators
        sentiment_components = pd.DataFrame(index=stock_data.index)
        sentiment_components['RSI_norm'] = normalize_series(stock_data['RSI'])
        sentiment_components['MACD_norm'] = normalize_series(stock_data['MACD'])
        sentiment_components['Price_MA20_norm'] = normalize_series(stock_data['Price_vs_MA20'])
        sentiment_components['Price_MA50_norm'] = normalize_series(stock_data['Price_vs_MA50'])
        
        # Inverse volatility (lower volatility = more positive sentiment)
        sentiment_components['Volatility_inv_norm'] = 1 - normalize_series(stock_data['Volatility_20d'])
        
        # Volume sentiment (bell curve around 1.5x average volume)
        volume_sentiment = np.exp(-((stock_data['Relative_Volume'] - 1.5)**2) / 0.5)
        sentiment_components['Volume_sentiment_norm'] = normalize_series(volume_sentiment)
        
        # Composite sentiment score (equally weighted, scaled 0-100)
        sentiment_scores = sentiment_components.mean(axis=1) * 100
        
        # Return the current (latest) sentiment score
        return sentiment_scores.iloc[-1] if not pd.isna(sentiment_scores.iloc[-1]) else np.nan
        
    except Exception as e:
        print(f"Error calculating sentiment: {e}")
        return np.nan

def analyze_multiple_stocks_sentiment(stock_pack, 
                                    periods=[10, 5, 2]):
    """
    Analyze multiple stocks' sentiment across different time periods
    
    Parameters:
    stock_pack (list): List of stock ticker symbols
    periods (list): List of years to analyze (default: [10, 5, 2])
    
    Returns:
    DataFrame: Sentiment analysis results with color coding
    """



    print(f"=== MULTIPLE STOCKS SENTIMENT ANALYSIS ===")
    print(f"Analyzing {len(stock_pack)} stocks: {', '.join(stock_pack)}")
    print(f"Time periods: {periods} years")
    print("=" * 60)
    
    # Initialize results dataframe
    results_df = pd.DataFrame(index=stock_pack)
    
    for period in periods:
        col_name = f"{period}_Year_Sentiment"
        results_df[col_name] = np.nan
    
    # Calculate sentiment for each stock and period
    for i, stock in enumerate(stock_pack):
        print(f"\nProcessing {stock} ({i+1}/{len(stock_pack)})...")
        
        for period in periods:
            try:
                # Download data for the specific period
                end_date = datetime.now()
                start_date = end_date - timedelta(days=int(period * 365))
                
                stock_data = yf.download(stock, start=start_date, end=end_date, progress=False)
                
                if stock_data.empty:
                    print(f"  No data available for {stock} ({period}Y)")
                    continue
                
                # Clean column structure
                if len(stock_data.columns.levels) > 1:
                    stock_data = stock_data.droplevel('Ticker', axis=1)
                
                stock_data.columns = ['Close', 'High', 'Low', 'Open', 'Volume']
                
                # Calculate sentiment score
                sentiment = calculate_sentiment_score(stock_data)
                results_df.loc[stock, f"{period}_Year_Sentiment"] = sentiment
                
                print(f"  {period}Y sentiment: {sentiment:.1f}" if not pd.isna(sentiment) else f"  {period}Y sentiment: N/A")
                
            except Exception as e:
                print(f"  Error processing {stock} ({period}Y): {e}")
                continue
    
    # Round the results for better display
    results_df = results_df.round(1)
    
    # Create visualization
    create_sentiment_heatmap(results_df, stock_pack, periods)
    
    # Display results table
    print(f"\n=== SENTIMENT ANALYSIS RESULTS ===")
    print("Higher values (red) = More bullish/greedy sentiment")
    print("Lower values (green) = More bearish/fearful sentiment")
    print("Range: 0-100 (Typical range: 20-80)")
    print()
    
    # Create a formatted display
    display_results_table(results_df)
    
    return results_df

def create_sentiment_heatmap(results_df, stock_pack, periods):
    """Create a color-coded heatmap of sentiment scores"""
    
    # Create figure
    fig, ax = plt.subplots(figsize=(12, max(8, len(stock_pack) * 0.5)))
    
    # Create custom colormap (green for low sentiment, red for high sentiment)
    # Using reversed RdYlGn colormap (Red-Yellow-Green reversed)
    cmap = plt.cm.RdYlGn_r
    
    # Create the heatmap
    im = ax.imshow(results_df.values, cmap=cmap, aspect='auto', vmin=20, vmax=80)
    
    # Set ticks and labels
    ax.set_xticks(range(len(periods)))
    ax.set_xticklabels([f"{p} Year" for p in periods])
    ax.set_yticks(range(len(stock_pack)))
    ax.set_yticklabels(stock_pack)
    
    # Add colorbar
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label('Sentiment Score (0-100)', rotation=270, labelpad=15)
    
    # Add text annotations
    for i in range(len(stock_pack)):
        for j in range(len(periods)):
            value = results_df.iloc[i, j]
            if not pd.isna(value):
                text = f'{value:.1f}'
                # Choose text color based on background
                text_color = 'white' if value < 40 or value > 70 else 'black'
                ax.text(j, i, text, ha="center", va="center", color=text_color, fontweight='bold')
            else:
                ax.text(j, i, 'N/A', ha="center", va="center", color='gray')
    
    # Styling
    ax.set_title(f'Stock Sentiment Analysis Heatmap\n(Green=Bearish/Fear, Red=Bullish/Greed)', 
                fontsize=14, fontweight='bold', pad=20)
    ax.set_xlabel('Time Period', fontsize=12)
    ax.set_ylabel('Stock Symbol', fontsize=12)
    
    plt.tight_layout()
    plt.show()

def display_results_table(results_df):
    """Display results in a formatted table with color indicators"""
    
    print(f"{'Stock':<8}", end="")
    periods = [col.replace('_Year_Sentiment', '') for col in results_df.columns]
    for period in periods:
        print(f"{period:>10}Y", end="")
    print()
    print("-" * (8 + len(periods) * 11))
    
    for stock in results_df.index:
        print(f"{stock:<8}", end="")
        for col in results_df.columns:
            value = results_df.loc[stock, col]
            if pd.isna(value):
                print(f"{'N/A':>11}", end="")
            else:
                # Add color indicator based on value
                if value >= 70:
                    indicator = "🔴"  # Red for high sentiment
                elif value >= 60:
                    indicator = "🟠"  # Orange for high-medium
                elif value >= 40:
                    indicator = "🟡"  # Yellow for neutral
                elif value >= 30:
                    indicator = "🟢"  # Green for low-medium
                else:
                    indicator = "🟢"  # Green for low sentiment
                
                print(f"{value:>8.1f}{indicator:>2}", end="")
        print()
    
    print("\nLegend:")
    print("🔴 Very Bullish (≥70) - Extreme Greed")
    print("🟠 Bullish (60-70) - Optimism") 
    print("🟡 Neutral (40-60) - Mixed Sentiment")
    print("🟢 Bearish (<40) - Fear/Caution")

def get_stock_info(stock_pack):
    """Get basic information about the stocks in the pack"""
    print(f"\n=== STOCK PACK INFORMATION ===")
    for stock in stock_pack:
        try:
            ticker = yf.Ticker(stock)
            info = ticker.info
            name = info.get('longName', 'N/A')
            sector = info.get('sector', 'N/A')
            print(f"{stock:<8}: {name[:40]:<40} | {sector}")
        except:
            print(f"{stock:<8}: Information not available")

# Usage Examples and Main Function
if __name__ == "__main__":
    
    # Example stock pack - you can modify this list
    stock_pack = AAA
    
    print("Multiple Stocks Sentiment Analysis Tool")
    hk_now = datetime.now(ZoneInfo("Asia/Hong_Kong"))
    ny_now = datetime.now(ZoneInfo("America/New_York"))

    print(f"HK time: {hk_now:%Y-%m-%d %H:%M:%S}")
    print(f"New York time: {ny_now:%Y-%m-%d %H:%M:%S}")
    print("-" * 50)
    
    # Show stock information
    get_stock_info(stock_pack)
    
    # Run the analysis
    results = analyze_multiple_stocks_sentiment(stock_pack=stock_pack, periods=[10, 5, 2])
    
    # Additional analysis
    print(f"\n--- SUMMARY STATISTICS ---")
    print("Average sentiment by period:")
    for col in results.columns:
        period = col.replace('_Year_Sentiment', '')
        avg_sentiment = results[col].mean()
        print(f"{period:>2}Y Average: {avg_sentiment:.1f}")
    
    print(f"\nMost bullish stocks (highest average sentiment):")
    avg_sentiment = results.mean(axis=1).sort_values(ascending=False)
    for i, (stock, avg) in enumerate(avg_sentiment.head(10).items()):
        print(f"{i+1}. {stock}: {avg:.1f}")
    
    print(f"\nMost bearish stocks (lowest average sentiment):")
    for i, (stock, avg) in enumerate(avg_sentiment.tail(10).iloc[::-1].items()):
        print(f"{i+1}. {stock}: {avg:.1f}")

# Quick analysis function
def quick_multi_analysis(stocks, periods=[10, 5, 2]):
    """Quick analysis function"""
    return analyze_multiple_stocks_sentiment(stock_pack=stocks, periods=periods)

# Example usage with custom stock pack:
# custom_stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA']
# results = analyze_multiple_stocks_sentiment(stock_pack=custom_stocks)
#------------------------------------Sentiment for Short + ETF -----------------------------
```

```python
# Correlation analysis: pairwise (TSLA/QQQ) and pack matrices
# Requirements: pandas, yfinance, matplotlib, seaborn
# Usage: python tsla_qqq_corr.py

import pandas as pd
import numpy as np
import datetime as dt
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf

# Characterize the two primary companies/tickers here
StockA, StockB = "TSLA", "QQQ"
START, END = '2015-01-01', dt.date.today().isoformat()
DEFAULT_WINDOWS = (30, 60, 90)

# ============== Pair utilities ==============

def fetch_prices(stock_a=StockA, stock_b=StockB, start=START, end=END):
    """Fetches and prepares close prices for two stocks."""
    px = yf.download([stock_a, stock_b], start=start, end=end, auto_adjust=True)['Close']
    return px.to_frame() if isinstance(px, pd.Series) else px[[stock_a, stock_b]].dropna()

def compute_returns(prices: pd.DataFrame):
    """Computes percentage change returns."""
    return prices.pct_change().dropna()

def rolling_corr_series(returns: pd.DataFrame, stock_a=StockA, stock_b=StockB, window: int = 60):
    """Computes a single rolling correlation series."""
    return returns[stock_a].rolling(window).corr(returns[stock_b]).dropna()

def plot_multi_window_corr(prices: pd.DataFrame, stock_a=StockA, stock_b=StockB, windows=DEFAULT_WINDOWS):
    """Plots rolling correlations for multiple windows on a single chart."""
    ret = compute_returns(prices)
    corr_df = pd.DataFrame({f'corr_{w}d': rolling_corr_series(ret, stock_a, stock_b, w) for w in windows}).dropna(how='all')
    
    fig, ax = plt.subplots(figsize=(10, 5))
    corr_df.plot(ax=ax, lw=1.5)
    ax.axhline(0, color='gray', lw=1, ls='--')
    win_str = "/".join(map(str, windows))
    ax.set(
        title=f'{stock_a} vs {stock_b} Rolling Daily Return Correlation ({win_str} Days)',
        xlabel='Date',
        ylabel='Correlation'
    )
    ax.grid(True, alpha=0.3)
    ax.legend(title='Window')
    plt.tight_layout()
    return fig, ax

# ============== Pack utilities and plotting ==============

def _download_close(tickers, start, end):
    """Downloads adjusted close prices for a list of tickers."""
    data = yf.download(tickers, start=start, end=end, auto_adjust=True)['Close']
    if isinstance(data, pd.Series): data = data.to_frame()
    return data[[c for c in tickers if c in data.columns]]

def compute_pack_correlations(stock_pack, years_list=(2, 5, 10), end_date=None):
    """Computes correlation matrices for multiple time horizons."""
    end_date = pd.to_datetime(end_date).date() if end_date else dt.date.today()
    results = {}
    for yrs in years_list:
        start_date = (pd.Timestamp(end_date) - pd.DateOffset(years=yrs)).date()
        px = _download_close(stock_pack, start_date.isoformat(), end_date.isoformat()).dropna(axis=1, how='all')
        ret = px.pct_change().dropna(how='all').dropna(axis=1, how='all')
        results[f'{yrs}y'] = ret.corr().clip(lower=-1, upper=1)
    return results

def get_top_corr_pairs(corr_matrix: pd.DataFrame, top_n: int = 5) -> pd.Series:
    """Return top-N correlated pairs with clear ticker labels."""
    # keep only upper triangle without diagonal
    mask = np.triu(np.ones(corr_matrix.shape, dtype=bool), k=1)
    upper = corr_matrix.where(mask)
    # stack to a Series with MultiIndex (row_ticker, col_ticker)
    pairs = upper.stack()
    pairs.index = pairs.index.set_names(["Ticker A", "Ticker B"])
    # sort high to low and return top N
    return pairs.sort_values(ascending=False).head(top_n)


def plot_heatmaps(corr_dict, order=("10y", "5y", "2y"), annot=True, fmt='.2f'):
    """Creates heatmaps and prints top correlated pairs for each time horizon."""
    figures, cmap = {}, 'RdYlGn_r'
    for key in order:
        corr = corr_dict.get(key)
        if corr is None or corr.empty:
            print(f"--- No data available for {key} correlation ---")
            continue
            
        fig, ax = plt.subplots(figsize=(max(8, corr.shape[1]), max(6, corr.shape[0] * 0.7)))
        mask = np.triu(np.ones_like(corr, dtype=bool))
        sns.heatmap(corr, mask=mask, vmin=-1, vmax=1, cmap=cmap, annot=annot, fmt=fmt, square=True, ax=ax)
        ax.set_title(f'{key} correlation')
        plt.tight_layout()
        figures[key] = fig
        
        print(f"--- Diversification Warning for {key} ---")
        print("Top 5 most correlated stock pairs:")
        print(get_top_corr_pairs(corr))
        print("-" * (35 + len(key)))
    return figures

if __name__ == '__main__':
    # Define the pack of stocks for correlation matrix analysis
    stock_pack = AAA
    
    # Compute and plot correlations for the stock pack
    corr_dict = compute_pack_correlations(stock_pack, years_list=(10, 5, 2))
    plot_heatmaps(corr_dict, order=("10y", "5y", "2y"))
    
    # Example for single-pair multi-window plot
    # prices = fetch_prices(StockA, StockB, start=START, end=END)
    # plot_multi_window_corr(prices)

    plt.show()

```
